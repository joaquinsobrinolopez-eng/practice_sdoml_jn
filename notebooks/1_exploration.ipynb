# Data exploration

In this notebook, some graphs regarding to the exploration analysis of the database are showed.

### 1. Initial load
Importing the diabetes dataset from `data/raw`. Also check dimensions of the dataset.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Global visualizing configuration
sns.set_theme(style="whitegrid")

# Loading data
data_path = Path.cwd().parent / "data" / "raw" / "diabetes_risk.csv"
df = pd.read_csv(data_path)

print(f"Dataset dimension: {df.shape}")
df.head()
```

### 2. Target class balance
Here, we analyse the distribution of the target variable to detect if there is a high class imbalance before training the dataset.

```python
plt.figure(figsize=(6, 4))
target_col = df.columns[-1]

sns.countplot(data=df, x=target_col, hue=target_col, palette="viridis", legend=False)
plt.title(f"Distribution of the classes: {target_col}", fontsize=12, fontweight="bold")
plt.xlabel("Class", fontsize=10)
plt.ylabel("Registers", fontsize=10)
plt.tight_layout()
plt.show()
```

In light of the results, we checked out that the distribution of the classes is desbalanced, with more low cases than high or moderated ones.

### 3. Distribution of continious variables
Inspection of symmetry and dispersion for continious variables to determine if those need any transformation.

```python
numeric_cols = df.select_dtypes(include=["number"]).columns
max_limit = len(numeric_cols)

for i in range(1, max_limit): # To avoid patient id 
    feature_col = numeric_cols[i]  # Selects the first variable (first variable)
    
    plt.figure(figsize=(7, 4))
    sns.histplot(data=df, x=feature_col, kde=True, color="teal", bins=25)
    plt.title(f"Distribution histogram: {feature_col}", fontsize=12, fontweight="bold")
    plt.xlabel(feature_col, fontsize=10)
    plt.ylabel("Densidad / Frecuencia", fontsize=10)
    plt.tight_layout()
    plt.show()
```

### 4. Outline detection and separation from classes.
Boxplots allow the identification of outliers and see which variables show more different among diagnoses

```python
numeric_cols = df.select_dtypes(include=["number"]).columns
target_col = df.columns[-1]

# Slicing de las columnas del índice 1 al 10
for feature_col in numeric_cols[1:11]:
    plt.figure(figsize=(7, 4))
    
    # legend=True activa la generación de elementos visuales en la leyenda
    ax = sns.boxplot(
        data=df, 
        x=target_col, 
        y=feature_col, 
        hue=target_col, 
        palette="Set2",
        legend=True
    )
    
    plt.title(f"Distribution of {feature_col} segmented by {target_col}", fontsize=12, fontweight="bold")
    plt.xlabel("Class", fontsize=10)
    plt.ylabel(feature_col, fontsize=10)
    
    # Reposiciona y titula la leyenda formalmente
    sns.move_legend(ax, "upper right", title="Diagnosis")
    
    plt.tight_layout()
    plt.show()
```

### 5. Lineal correlation analysis
Redundancies or dependencies among the numerical variables via lineal correlation.

```python
plt.figure(figsize=(12, 10))
corr = df.select_dtypes(include=["number"]).corr()

sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, square=True)
plt.title("Matriz de Correlación entre Variables Numéricas", fontsize=12, fontweight="bold")
plt.xlabel("Características", fontsize=10)
plt.ylabel("Características", fontsize=10)
plt.tight_layout()
plt.show()
```
